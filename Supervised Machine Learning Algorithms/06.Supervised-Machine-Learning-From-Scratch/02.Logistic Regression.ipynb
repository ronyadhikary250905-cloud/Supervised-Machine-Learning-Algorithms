```python
import numpy as np
```

```python
# Logistic Regression

class LogisticRegression:
    def __init__(self, learning_rate=0.1, n_iter=1000):
        self.bias = None
        self.weights = None
        self.lr = learning_rate
        self.n_iter = n_iter

    def _sigmoid(self, z):
        return (1 / (1 + np.exp(-z)))

    def fit(self, X, y):
        m, n = X.shape

        self.bias = 0
        self.weights = np.zeros(n)

        for i in range(self.n_iter):
            z = self.bias + np.dot(X, self.weights)
            y_pred = self._sigmoid(z)

            db = (1/m) * (y_pred - y)
            dw = (1/m) * np.dot(X.T, (y_pred - y))
            
            self.bias -= self.lr * db
            self.weights -= self.lr * dw
    
    def get_probabilities(self, X):
        z = self.bias + np.dot(X, self.weights)
        return self._sigmoid(z)
        
    def predict(self, X, threshold=0.5):
        probabilities = self.get_probabilities(X)
        y_pred_bool = probabilities >= threshold
        return y_pred_bool.astype(int)
```

```python
X = np.array([[1, 2], [2, 3], [3, 4], [4, 5]])
y = np.array([0, 0, 1, 1])

model = LogisticRegression()
model.fit(X, y)

y_pred = model.predict(X)
print("Predictions:", y_pred)

print("Probabilities:",model.get_probabilities(X))
```

```python
# Estimates model parameters by maximizing likelihood:

# Dataset
X = np.array([[1, 2], [2, 3], [3, 4], [4, 5]])
y = np.array([0, 0, 1, 1])

# Initialize weights and bias
weights = np.zeros(X.shape[1])
bias = 0.0

learning_rate = 0.1
epochs = 5000
m = len(y)

# 1. Sigmoid Function
def sigmoid(z):
    z = np.clip(z, -500, 500)
    return 1 / (1 + np.exp(-z))

# 2. Training using MLE (Gradient Descent)
for i in range(epochs):

    # Linear equation
    z = np.dot(X, weights) + bias

    # Calculate probabilities
    y_prob = sigmoid(z)

    # Calculate gradients
    dw = np.dot(X.T, (y_prob - y)) / m
    db = np.sum(y_prob - y) / m

    # Update weights and bias
    weights -= learning_rate * dw
    bias -= learning_rate * db

# 3. Prediction probabilities
z = np.dot(X, weights) + bias
probabilities = sigmoid(z)

# 4. Predict class labels
threshold = 0.5
y_pred = (probabilities >= threshold).astype(int)
```

```python

print("Predictions:", y_pred)

print("\nProbabilities:")
print(np.column_stack((1 - probabilities, probabilities)))

print("\nWeights:", weights)
print("Bias:", bias)
```
